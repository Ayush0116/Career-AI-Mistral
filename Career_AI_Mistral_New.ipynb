```python
def update_skill_progress(
    degree,
    field,
    role,
    skill,
    assessment_score=0,
    assessment_completed=False,
    error_challenge_completed=False,
    practice_score=0
):
    skill_data = career_data[degree][field][role]["skill_progress"][skill]

    if assessment_completed:
        skill_data["assessments_completed"] += 1

    if error_challenge_completed:
        skill_data["error_challenges_completed"] += 1

    skill_data["assessment_score"] = assessment_score
    skill_data["practice_score"] = practice_score

    # Assessment + practical performance
    assessment_progress = assessment_score
    practical_progress = practice_score

    skill_data["progress_percentage"] = round(
        (assessment_progress * 0.6) +
        (practical_progress * 0.4)
    )

    if skill_data["progress_percentage"] == 0:
        skill_data["status"] = "Not Started"
    elif skill_data["progress_percentage"] < 40:
        skill_data["status"] = "Needs Practice"
    elif skill_data["progress_percentage"] < 70:
        skill_data["status"] = "Learning"
    elif skill_data["progress_percentage"] < 90:
        skill_data["status"] = "Good"
    else:
        skill_data["status"] = "Ready"

    return skill_data


# Example
result = update_skill_progress(
    degree="B.Tech",
    field="Computer Science & Engineering",
    role="Software Developer",
    skill="C++",
    assessment_score=75,
    assessment_completed=True,
    error_challenge_completed=True,
    practice_score=65
)

print("Skill progress updated!")
print(result)
```

```python
def generate_dynamic_questions(skill, role, current_progress, difficulty="medium", num_questions=5):
    prompt = f"""
You are an expert technical assessment generator.

Target Role: {role}
Skill: {skill}
Current Skill Progress: {current_progress}%
Difficulty: {difficulty}

Generate exactly {num_questions} NEW assessment questions for this skill.

Rules:
- Questions must be directly related to the selected skill.
- Do not repeat common/basic questions unnecessarily.
- Match the difficulty with the user's current progress.
- Include practical and conceptual questions.
- Do not provide answers.
- Return only numbered questions.
"""

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=500,
            temperature=0.8,
            top_p=0.9,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    generated_text = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    return generated_text.strip()


# Test
questions = generate_dynamic_questions(
    skill="C++",
    role="Software Developer",
    current_progress=35,
    difficulty="medium",
    num_questions=5
)

print("Generated Questions:\n")
print(questions)
```

```python
def evaluate_assessment(skill, role, questions, user_answers):
    prompt = f"""
You are an expert technical evaluator.

Target Role: {role}
Skill: {skill}

Questions:
{questions}

User Answers:
{user_answers}

Evaluate the user's answers.

For each question provide:
1. Score out of 20
2. Correctness
3. What was done well
4. What is missing or incorrect
5. How to improve

At the end provide:
- Total Score out of 100
- Skill Performance Percentage
- Strengths
- Weak Areas
- Recommended Practice Topics

Be technically accurate and do not give unnecessary information.
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=4096
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=1000,
            temperature=0.3,
            top_p=0.9,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    result = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    return result.strip()


print("Assessment evaluation function created successfully!")
```

```python
def apply_assessment_result(
    degree,
    field,
    role,
    skill,
    score_percentage
):
    skill_data = career_data[degree][field][role]["skill_progress"][skill]

    # Mark skill as selected
    skill_data["selected_by_user"] = True

    # Update assessment score
    skill_data["assessment_score"] = score_percentage

    # Calculate current progress
    skill_data["progress_percentage"] = round(
        (score_percentage * 0.6) +
        (skill_data["practice_score"] * 0.4)
    )

    # Update status
    progress = skill_data["progress_percentage"]

    if progress < 40:
        skill_data["status"] = "Needs Practice"
    elif progress < 70:
        skill_data["status"] = "Learning"
    elif progress < 90:
        skill_data["status"] = "Good"
    else:
        skill_data["status"] = "Ready"

    return skill_data


# Example
updated_skill = apply_assessment_result(
    degree="B.Tech",
    field="Computer Science & Engineering",
    role="Software Developer",
    skill="C++",
    score_percentage=78
)

print("Skill progress updated successfully!")
print(updated_skill)
```

```python
def generate_skill_gap_analysis(degree, field, role):
    role_data = career_data[degree][field][role]

    skill_progress = role_data["skill_progress"]

    progress_data = []

    for skill, data in skill_progress.items():
        progress_data.append({
            "skill": skill,
            "selected": data["selected_by_user"],
            "progress": data["progress_percentage"],
            "status": data["status"],
            "assessment_score": data["assessment_score"],
            "practice_score": data["practice_score"]
        })

    prompt = f"""
You are an AI career analysis engine.

Degree: {degree}
Field: {field}
Target Role: {role}

Current Skill Progress:
{progress_data}

Analyze the user's current readiness for the target role.

Return:

1. Strong Skills
2. Skills Needing Improvement
3. Critical Skill Gaps
4. Skills to Practice First
5. Recommended Learning Order
6. Current Role Readiness Percentage
7. Short Explanation of the Readiness

Rules:
- Base the analysis only on the provided progress data.
- Do not assume that an untested skill is mastered.
- Prioritize skills with low performance.
- Consider both assessment and practical performance.
- The analysis must change when the user's progress changes.
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=6000
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=1200,
            temperature=0.4,
            top_p=0.9,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    result = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    return result.strip()


# Test
analysis = generate_skill_gap_analysis(
    degree="B.Tech",
    field="Computer Science & Engineering",
    role="Software Developer"
)

print("Dynamic Skill Gap Analysis:\n")
print(analysis)
```

```python
def generate_personalized_roadmap(
    degree,
    field,
    role,
    timeline_months
):
    role_data = career_data[degree][field][role]

    progress_data = []

    for skill, data in role_data["skill_progress"].items():
        progress_data.append({
            "skill": skill,
            "progress": data["progress_percentage"],
            "status": data["status"],
            "assessment_score": data["assessment_score"],
            "practice_score": data["practice_score"]
        })

    prompt = f"""
You are an AI career roadmap generator.

Degree: {degree}
Field: {field}
Target Role: {role}
Available Preparation Time: {timeline_months} months

Current Skill Progress:
{progress_data}

Create a personalized preparation roadmap.

Requirements:

1. Divide the roadmap into {timeline_months} months.
2. Prioritize weak and critical skills first.
3. Do not spend unnecessary time on skills where the user already performs well.
4. Include learning, practice, assessments, and practical projects.
5. Include error-solving/debugging practice where relevant.
6. Recommend revision for weak skills.
7. Define measurable goals for every month.
8. The roadmap must be realistic for the available timeline.
9. Do not ask the user to manually enter their skills again.
10. Adapt the roadmap according to the current progress data.

Return:

- Current readiness
- Month-wise roadmap
- Skills to learn
- Skills to improve
- Practice tasks
- Assessment plan
- Project milestones
- Final preparation stage
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=7000
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=1800,
            temperature=0.5,
            top_p=0.9,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    result = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    return result.strip()


# Test with a 6-month preparation timeline
roadmap = generate_personalized_roadmap(
    degree="B.Tech",
    field="Computer Science & Engineering",
    role="Software Developer",
    timeline_months=6
)

print("Personalized Roadmap:\n")
print(roadmap)
```

```python
user_profile = {
    "degree": None,
    "field": None,
    "target_role": None,
    "timeline_months": None,
    "selected_skills": [],
    "skill_progress": {},
    "overall_progress": 0,
    "readiness_percentage": 0
}

def create_user_profile(
    degree,
    field,
    target_role,
    timeline_months,
    selected_skills
):
    role_data = career_data[degree][field][target_role]

    profile = {
        "degree": degree,
        "field": field,
        "target_role": target_role,
        "timeline_months": timeline_months,
        "selected_skills": selected_skills,
        "skill_progress": {},
        "overall_progress": 0,
        "readiness_percentage": 0
    }

    for skill in selected_skills:
        if skill not in role_data["skill_progress"]:
            raise ValueError(f"Skill not found for role: {skill}")

        profile["skill_progress"][skill] = {
            "progress_percentage": 0,
            "assessment_score": 0,
            "practice_score": 0,
            "assessments_completed": 0,
            "error_challenges_completed": 0,
            "status": "Not Started"
        }

    return profile


print("User profile system created successfully!")
```

```python
def calculate_overall_progress(profile):
    skill_progress = profile["skill_progress"]

    if not skill_progress:
        profile["overall_progress"] = 0
        profile["readiness_percentage"] = 0
        return profile

    total_progress = sum(
        skill["progress_percentage"]
        for skill in skill_progress.values()
    )

    overall_progress = round(
        total_progress / len(skill_progress)
    )

    profile["overall_progress"] = overall_progress

    # Readiness currently based on demonstrated skill performance.
    profile["readiness_percentage"] = overall_progress

    return profile


# Example profile
profile = create_user_profile(
    degree="B.Tech",
    field="Computer Science & Engineering",
    target_role="Software Developer",
    timeline_months=6,
    selected_skills=[
        "C++",
        "Data Structures",
        "SQL",
        "JavaScript"
    ]
)

# Example: user has progressed in some skills
profile["skill_progress"]["C++"]["progress_percentage"] = 75
profile["skill_progress"]["Data Structures"]["progress_percentage"] = 50
profile["skill_progress"]["SQL"]["progress_percentage"] = 80
profile["skill_progress"]["JavaScript"]["progress_percentage"] = 40

profile = calculate_overall_progress(profile)

print("Overall Progress:", profile["overall_progress"], "%")
print("Role Readiness:", profile["readiness_percentage"], "%")
```

```python
def evaluate_skill_assessment(
    user_state,
    skill,
    assessment_id,
    user_answers
):
    skill_data = user_state["skill_progress"][skill]

    # Find the requested assessment
    assessment = None

    for item in skill_data["assessments"]:
        if item["assessment_id"] == assessment_id:
            assessment = item
            break

    if assessment is None:
        raise ValueError("Assessment not found.")

    questions = assessment["questions"]

    prompt = f"""
You are an expert technical evaluator.

Target Role: {user_state["target_role"]}
Skill: {skill}

Assessment Questions:
{questions}

User Answers:
{user_answers}

Evaluate the user's answers carefully.

For each question provide:
- Score out of 20
- Correctness
- What was done well
- What is incorrect or missing
- How to improve

Then provide:

TOTAL SCORE: X/100
PERFORMANCE: X%

Also provide:
STRENGTHS:
WEAK AREAS:
RECOMMENDED PRACTICE:

Important:
- Evaluate only the answers provided.
- Do not assume an unanswered question is correct.
- Be technically accurate.
- Do not give full marks without sufficient justification.
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=6000
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=1500,
            temperature=0.3,
            top_p=0.9,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    result = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    ).strip()

    # Save answers and evaluation
    assessment["answers"] = user_answers
    assessment["evaluation"] = result
    assessment["evaluated"] = True

    return result


# Example:
# User ke actual answers yahan frontend se aayenge.
# Abhi sirf testing ke liye example structure:

user_answers = [
    "My answer to question 1",
    "My answer to question 2",
    "My answer to question 3",
    "My answer to question 4",
    "My answer to question 5"
]

evaluation = evaluate_skill_assessment(
    user_state=user_state,
    skill="C++",
    assessment_id=1,
    user_answers=user_answers
)

print(evaluation)
```

```python
import re


def extract_performance_score(evaluation_text):
    """
    Mistral ke evaluation output se PERFORMANCE percentage extract karta hai.
    """

    patterns = [
        r"PERFORMANCE\s*:\s*(\d+(?:\.\d+)?)\s*%",
        r"Performance\s*:\s*(\d+(?:\.\d+)?)\s*%",
        r"TOTAL SCORE\s*:\s*(\d+(?:\.\d+)?)\s*/\s*100"
    ]

    for pattern in patterns:
        match = re.search(pattern, evaluation_text)

        if match:
            score = float(match.group(1))

            # Score ko 0-100 range me rakho
            score = max(0, min(100, score))

            return round(score)

    raise ValueError(
        "Mistral evaluation me valid performance score nahi mila."
    )


def update_skill_from_assessment(
    user_state,
    skill,
    assessment_id,
    evaluation_text
):
    """
    Assessment evaluation ke basis par skill progress automatically update karta hai.
    """

    skill_data = user_state["skill_progress"][skill]

    # Mistral evaluation se score nikalo
    score = extract_performance_score(evaluation_text)

    # Assessment find karo
    assessment = None

    for item in skill_data["assessments"]:
        if item["assessment_id"] == assessment_id:
            assessment = item
            break

    if assessment is None:
        raise ValueError("Assessment not found.")

    # Assessment ka score save karo
    assessment["score"] = score
    assessment["evaluated"] = True

    # Completed assessment count update
    skill_data["assessments_completed"] += 1

    # Latest assessment score
    skill_data["assessment_score"] = score

    # Assessment + practical progress
    assessment_progress = score
    practice_progress = skill_data["practice_score"]

    # Automatic weighted progress
    skill_data["progress_percentage"] = round(
        (assessment_progress * 0.6) +
        (practice_progress * 0.4)
    )

    # Status automatically calculate
    progress = skill_data["progress_percentage"]

    if progress == 0:
        skill_data["status"] = "Not Started"
    elif progress < 40:
        skill_data["status"] = "Needs Practice"
    elif progress < 70:
        skill_data["status"] = "Learning"
    elif progress < 90:
        skill_data["status"] = "Good"
    else:
        skill_data["status"] = "Ready"

    return skill_data


# Example:
# Evaluation jo Step 22 se mila tha usko use karo.

updated_skill = update_skill_from_assessment(
    user_state=user_state,
    skill="C++",
    assessment_id=1,
    evaluation_text=evaluation
)

print("Skill:", "C++")
print("Assessment Score:", updated_skill["assessment_score"], "%")
print("Skill Progress:", updated_skill["progress_percentage"], "%")
print("Status:", updated_skill["status"])
print(
    "Assessments Completed:",
    updated_skill["assessments_completed"]
)
```

```python
def calculate_cumulative_skill_progress(
    user_state,
    skill
):
    skill_data = user_state["skill_progress"][skill]

    completed_assessments = [
        assessment
        for assessment in skill_data.get("assessments", [])
        if assessment.get("evaluated") is True
        and assessment.get("score") is not None
    ]

    if not completed_assessments:
        skill_data["assessment_score"] = 0
        skill_data["progress_percentage"] = 0
        skill_data["status"] = "Not Started"
        return skill_data

    # All completed assessment scores
    scores = [
        assessment["score"]
        for assessment in completed_assessments
    ]

    # Cumulative assessment performance
    cumulative_score = round(
        sum(scores) / len(scores)
    )

    skill_data["assessment_score"] = cumulative_score
    skill_data["assessments_completed"] = len(
        completed_assessments
    )

    # Practical progress comes from practice/error challenges
    practice_progress = skill_data["practice_score"]

    # Final skill progress
    skill_data["progress_percentage"] = round(
        (cumulative_score * 0.6) +
        (practice_progress * 0.4)
    )

    progress = skill_data["progress_percentage"]

    if progress == 0:
        skill_data["status"] = "Not Started"
    elif progress < 40:
        skill_data["status"] = "Needs Practice"
    elif progress < 70:
        skill_data["status"] = "Learning"
    elif progress < 90:
        skill_data["status"] = "Good"
    else:
        skill_data["status"] = "Ready"

    return skill_data


# Calculate current C++ progress
updated_skill = calculate_cumulative_skill_progress(
    user_state,
    "C++"
)

print("Skill:", "C++")
print("Assessments Completed:",
      updated_skill["assessments_completed"])
print("Assessment Performance:",
      updated_skill["assessment_score"], "%")
print("Overall Skill Progress:",
      updated_skill["progress_percentage"], "%")
print("Status:",
      updated_skill["status"])
```

```python
def calculate_user_overall_progress(user_state):
    skill_progress = user_state["skill_progress"]

    if not skill_progress:
        user_state["overall_progress"] = 0
        user_state["readiness_percentage"] = 0
        return user_state

    total_progress = sum(
        data["progress_percentage"]
        for data in skill_progress.values()
    )

    overall_progress = round(
        total_progress / len(skill_progress)
    )

    user_state["overall_progress"] = overall_progress
    user_state["readiness_percentage"] = overall_progress

    return user_state


# Calculate current overall progress
user_state = calculate_user_overall_progress(user_state)

print("Overall Role Progress:",
      user_state["overall_progress"], "%")

print("Role Readiness:",
      user_state["readiness_percentage"], "%")

print("\nSkill-wise Progress:")

for skill, data in user_state["skill_progress"].items():
    print(
        f"{skill}: "
        f"{data['progress_percentage']}% "
        f"({data['status']})"
    )
```

```python
def evaluate_submitted_attempt(
    user_state,
    skill,
    assessment_id,
    attempt_id
):
    skill_data = user_state["skill_progress"][skill]

    # Assessment find karo
    assessment = None

    for item in skill_data.get("assessments", []):
        if item["assessment_id"] == assessment_id:
            assessment = item
            break

    if assessment is None:
        raise ValueError("Assessment not found.")

    # Attempt find karo
    attempt = None

    for item in assessment.get("attempts", []):
        if item["attempt_id"] == attempt_id:
            attempt = item
            break

    if attempt is None:
        raise ValueError("Assessment attempt not found.")

    if not attempt["submitted"]:
        raise ValueError(
            "Assessment must be submitted before evaluation."
        )

    questions = assessment["questions"]
    answers = attempt["answers"]

    prompt = f"""
You are an expert technical evaluator.

Target Role: {user_state["target_role"]}
Skill: {skill}

QUESTIONS:
{questions}

USER ANSWERS:
{answers}

Evaluate all 5 answers carefully.

For each question provide:
- Score out of 20
- Correctness
- What was done well
- What is missing or incorrect
- How to improve

Then calculate the total score.

Return EXACTLY:

TOTAL SCORE: X/100
PERFORMANCE: X%

STRENGTHS:
- ...

WEAK AREAS:
- ...

RECOMMENDED PRACTICE:
- ...

Rules:
- Evaluate only the submitted answers.
- Do not assume missing information is correct.
- Score fairly.
- Be technically accurate.
- TOTAL SCORE and PERFORMANCE must represent the same percentage.
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=7000
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=1500,
            temperature=0.2,
            top_p=0.9,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    evaluation = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    ).strip()

    # Score automatically extract karo
    score = extract_performance_score(evaluation)

    # Save evaluation in this specific attempt
    attempt["score"] = score
    attempt["evaluation"] = evaluation

    # Assessment ka latest result
    assessment["score"] = score
    assessment["evaluation"] = evaluation
    assessment["evaluated"] = True

    return attempt


# Evaluate submitted attempt
evaluated_attempt = evaluate_submitted_attempt(
    user_state=user_state,
    skill=next_skill,
    assessment_id=next_assessment["assessment_id"],
    attempt_id=attempt["attempt_id"]
)

print("Assessment Evaluated Successfully")
print("Skill:", next_skill)
print("Attempt:", evaluated_attempt["attempt_id"])
print("Score:", evaluated_attempt["score"], "%")
print("\nEvaluation:\n")
print(evaluated_attempt["evaluation"])
```

```python
def update_skill_progress_from_attempts(
    user_state,
    skill,
    assessment_id
):
    skill_data = user_state["skill_progress"][skill]

    # Assessment find karo
    assessment = None

    for item in skill_data.get("assessments", []):
        if item["assessment_id"] == assessment_id:
            assessment = item
            break

    if assessment is None:
        raise ValueError("Assessment not found.")

    # Sirf evaluated attempts
    completed_attempts = [
        attempt
        for attempt in assessment.get("attempts", [])
        if attempt.get("score") is not None
    ]

    if not completed_attempts:
        return skill_data

    # Current assessment ke attempts ka average
    assessment_scores = [
        attempt["score"]
        for attempt in completed_attempts
    ]

    assessment_average = round(
        sum(assessment_scores) / len(assessment_scores)
    )

    # Skill ke sabhi assessments ke evaluated attempts
    all_scores = []

    for item in skill_data.get("assessments", []):
        for attempt in item.get("attempts", []):
            if attempt.get("score") is not None:
                all_scores.append(attempt["score"])

    if all_scores:
        cumulative_assessment_score = round(
            sum(all_scores) / len(all_scores)
        )
    else:
        cumulative_assessment_score = 0

    skill_data["assessment_score"] = cumulative_assessment_score
    skill_data["assessments_completed"] = len(all_scores)

    # Existing practical/debugging performance
    practice_score = skill_data["practice_score"]

    # Final skill progress
    skill_data["progress_percentage"] = round(
        (cumulative_assessment_score * 0.6) +
        (practice_score * 0.4)
    )

    progress = skill_data["progress_percentage"]

    if progress == 0:
        skill_data["status"] = "Not Started"
    elif progress < 40:
        skill_data["status"] = "Needs Practice"
    elif progress < 70:
        skill_data["status"] = "Learning"
    elif progress < 90:
        skill_data["status"] = "Good"
    else:
        skill_data["status"] = "Ready"

    # Progress history save karo
    save_skill_progress_history(
        user_state,
        skill
    )

    return skill_data


# Update C++ skill progress
updated_skill = update_skill_progress_from_attempts(
    user_state=user_state,
    skill=next_skill,
    assessment_id=next_assessment["assessment_id"]
)

print("Skill:", next_skill)
print(
    "Cumulative Assessment Score:",
    updated_skill["assessment_score"],
    "%"
)
print(
    "Practice Score:",
    updated_skill["practice_score"],
    "%"
)
print(
    "Overall Skill Progress:",
    updated_skill["progress_percentage"],
    "%"
)
print(
    "Status:",
    updated_skill["status"]
)
```

```python
import re


def extract_next_skill(next_skill_result, user_state):

    match = re.search(
        r"NEXT_SKILL\s*:\s*(.+)",
        next_skill_result,
        re.IGNORECASE
    )

    if not match:
        raise ValueError("Mistral response me NEXT_SKILL nahi mila.")

    next_skill = match.group(1).strip()

    # Exact skill match
    if next_skill not in user_state["skill_progress"]:
        raise ValueError(
            f"Mistral ne invalid skill select ki: {next_skill}"
        )

    return next_skill


def generate_next_skill_assessment(user_state, difficulty="medium"):

    priority_result = user_state.get("next_skill_priority", "")

    if not priority_result:
        raise ValueError(
            "Pehle get_mistral_next_skill(user_state) run karo."
        )

    next_skill = extract_next_skill(
        priority_result,
        user_state
    )

    current_progress = user_state[
        "skill_progress"
    ][next_skill]["progress_percentage"]

    role = user_state["target_role"]

    prompt = f"""
You are an expert technical assessment generator.

Target Role:
{role}

Skill:
{next_skill}

Current Progress:
{current_progress}%

Difficulty:
{difficulty}

Generate ONE completely NEW assessment.

The assessment must contain EXACTLY 5 questions.

Requirements:
- Questions must be different from previous assessments.
- Questions must match the target role.
- Questions must test practical understanding.
- Include conceptual and problem-solving questions.
- Do not provide answers.
- Do not invent the user's performance.
- Adapt difficulty according to current progress.

Return exactly:

ASSESSMENT:
1. ...
2. ...
3. ...
4. ...
5. ...
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=5000
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=900,
            temperature=0.85,
            top_p=0.9,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    questions = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    ).strip()

    # Save assessment
    skill_data = user_state["skill_progress"][next_skill]

    if "assessments" not in skill_data:
        skill_data["assessments"] = []

    assessment_id = len(skill_data["assessments"]) + 1

    assessment = {
        "assessment_id": assessment_id,
        "skill": next_skill,
        "difficulty": difficulty,
        "progress_at_generation": current_progress,
        "questions": questions,
        "attempts": [],
        "evaluated": False
    }

    skill_data["assessments"].append(assessment)

    return assessment


# Generate next fresh assessment
next_assessment = generate_next_skill_assessment(
    user_state,
    difficulty="medium"
)

print("Skill:", next_assessment["skill"])
print("Assessment ID:", next_assessment["assessment_id"])
print("\nQuestions:\n")
print(next_assessment["questions"])
```

```python
import json
import torch
import re


def evaluate_next_assessment(
    user_state,
    skill,
    assessment_id,
    attempt_id,
    user_answers
):

    if skill not in user_state["skill_progress"]:
        raise ValueError(f"Skill not found: {skill}")

    if len(user_answers) != 5:
        raise ValueError("Exactly 5 answers required hain.")

    skill_data = user_state["skill_progress"][skill]

    assessment = None

    for item in skill_data.get("assessments", []):
        if item["assessment_id"] == assessment_id:
            assessment = item
            break

    if assessment is None:
        raise ValueError("Assessment not found.")

    attempt = None

    for item in assessment.get("attempts", []):
        if item["attempt_id"] == attempt_id:
            attempt = item
            break

    if attempt is None:
        raise ValueError("Attempt not found.")

    questions = assessment["questions"]

    prompt = f"""
You are an expert technical evaluator.

Target Role:
{user_state["target_role"]}

Skill:
{skill}

Assessment Questions:
{questions}

User Answers:
{json.dumps(user_answers, indent=2)}

Evaluate all 5 answers carefully.

Give marks out of 20 for each question.

Total marks = 100.

Evaluate based on:
- Technical correctness
- Understanding
- Problem solving
- Practical application
- Accuracy

Do not give marks merely because an answer is long.

Return EXACTLY:

Q1_SCORE: X/20
Q2_SCORE: X/20
Q3_SCORE: X/20
Q4_SCORE: X/20
Q5_SCORE: X/20

TOTAL SCORE: X/100
PERFORMANCE: X%

STRENGTHS:
- ...

WEAK_AREAS:
- ...

EXPLANATION:
...

NEXT_PRACTICE:
...
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=8000
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=1200,
            temperature=0.3,
            top_p=0.9,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    evaluation = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    ).strip()

    # Score extract karo
    match = re.search(
        r"PERFORMANCE\s*:\s*(\d+(?:\.\d+)?)\s*%",
        evaluation,
        re.IGNORECASE
    )

    if not match:
        raise ValueError(
            "Mistral response me valid PERFORMANCE score nahi mila."
        )

    score = round(
        max(0, min(100, float(match.group(1))))
    )

    # Save attempt
    attempt["answers"] = user_answers
    attempt["submitted"] = True
    attempt["score"] = score
    attempt["evaluation"] = evaluation

    # Save latest assessment result
    assessment["score"] = score
    assessment["evaluation"] = evaluation
    assessment["evaluated"] = True

    return evaluation


# -------------------------------------------------
# Example:
# User ke actual 5 answers yahan frontend se aayenge
# -------------------------------------------------

user_answers = [
    "User answer 1",
    "User answer 2",
    "User answer 3",
    "User answer 4",
    "User answer 5"
]


evaluation = evaluate_next_assessment(
    user_state=user_state,
    skill=next_assessment["skill"],
    assessment_id=next_assessment["assessment_id"],
    attempt_id=attempt["attempt_id"],
    user_answers=user_answers
)

print(evaluation)
```

```python
def update_skill_after_assessment(
    user_state,
    skill,
    assessment_id,
    attempt_id
):
    if skill not in user_state["skill_progress"]:
        raise ValueError(f"Skill not found: {skill}")

    skill_data = user_state["skill_progress"][skill]

    assessment = None

    for item in skill_data.get("assessments", []):
        if item["assessment_id"] == assessment_id:
            assessment = item
            break

    if assessment is None:
        raise ValueError("Assessment not found.")

    attempt = None

    for item in assessment.get("attempts", []):
        if item["attempt_id"] == attempt_id:
            attempt = item
            break

    if attempt is None:
        raise ValueError("Attempt not found.")

    if attempt["score"] is None:
        raise ValueError("Assessment abhi evaluate nahi hua.")

    # -----------------------------------------
    # Sab completed attempts ke scores
    # -----------------------------------------

    all_scores = []

    for item in skill_data.get("assessments", []):
        for current_attempt in item.get("attempts", []):
            if current_attempt.get("score") is not None:
                all_scores.append(current_attempt["score"])

    # Average assessment performance
    assessment_score = round(
        sum(all_scores) / len(all_scores)
    ) if all_scores else 0

    skill_data["assessment_score"] = assessment_score
    skill_data["assessments_completed"] = len(all_scores)

    # -----------------------------------------
    # Assessment + debugging/practice progress
    # -----------------------------------------

    practice_score = skill_data.get("practice_score", 0)

    skill_data["progress_percentage"] = round(
        (assessment_score * 0.6) +
        (practice_score * 0.4)
    )

    # -----------------------------------------
    # Status
    # -----------------------------------------

    progress = skill_data["progress_percentage"]

    if progress == 0:
        status = "Not Started"
    elif progress < 40:
        status = "Needs Improvement"
    elif progress < 70:
        status = "Developing"
    elif progress < 90:
        status = "Good"
    else:
        status = "Ready"

    skill_data["status"] = status

    # -----------------------------------------
    # Overall user progress
    # -----------------------------------------

    refresh_user_progress(user_state)

    return skill_data


# Update skill progress
updated_skill = update_skill_after_assessment(
    user_state=user_state,
    skill=next_assessment["skill"],
    assessment_id=next_assessment["assessment_id"],
    attempt_id=attempt["attempt_id"]
)


print("Skill:", next_assessment["skill"])
print("Assessment Score:", updated_skill["assessment_score"], "%")
print("Practice Score:", updated_skill["practice_score"], "%")
print("Skill Progress:", updated_skill["progress_percentage"], "%")
print("Status:", updated_skill["status"])

print("\nOverall Progress:", user_state["overall_progress"], "%")
print("Readiness:", user_state["readiness_percentage"], "%")
```

```python
def generate_debugging_challenge(user_state, skill, difficulty="medium"):

    if skill not in user_state["skill_progress"]:
        raise ValueError(f"Skill not found: {skill}")

    skill_data = user_state["skill_progress"][skill]

    current_progress = skill_data["progress_percentage"]
    role = user_state["target_role"]

    prompt = f"""
You are an expert technical interviewer and debugging challenge generator.

Target Role:
{role}

Skill:
{skill}

Current Skill Progress:
{current_progress}%

Difficulty:
{difficulty}

Generate ONE completely NEW and realistic debugging challenge.

Requirements:
- Challenge must be directly related to the skill.
- Challenge must be relevant to the target role.
- Give realistic faulty code or technical problem.
- The error should require actual problem solving.
- Do NOT provide the solution.
- Do NOT repeat a previous challenge.
- Difficulty should match the user's current progress.
- User will submit their own solution later.

Return EXACTLY:

TITLE:
<challenge title>

LANGUAGE:
<programming language or technology>

PROBLEM:
<explain the situation>

FAULTY_CODE:
<faulty code>

TASK:
<what the user must fix>

EXPECTED_BEHAVIOR:
<what the corrected program/system should do>
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=6000
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=1000,
            temperature=0.85,
            top_p=0.9,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    challenge = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    ).strip()

    # Save challenge in user state
    if "debugging_challenges" not in skill_data:
        skill_data["debugging_challenges"] = []

    challenge_id = len(
        skill_data["debugging_challenges"]
    ) + 1

    challenge_data = {
        "challenge_id": challenge_id,
        "skill": skill,
        "difficulty": difficulty,
        "progress_at_generation": current_progress,
        "challenge": challenge,
        "user_solution": None,
        "score": None,
        "evaluation": None,
        "completed": False
    }

    skill_data["debugging_challenges"].append(
        challenge_data
    )

    return challenge_data


# -----------------------------------------
# Generate a NEW debugging challenge
# -----------------------------------------

debugging_challenge = generate_debugging_challenge(
    user_state=user_state,
    skill=next_assessment["skill"],
    difficulty="medium"
)

print("Skill:", debugging_challenge["skill"])
print("Challenge ID:", debugging_challenge["challenge_id"])
print("\n")
print(debugging_challenge["challenge"])
```

```python
def update_skill_after_debugging(
    user_state,
    skill,
    challenge_id
):

    if skill not in user_state["skill_progress"]:
        raise ValueError(f"Skill not found: {skill}")

    skill_data = user_state["skill_progress"][skill]

    challenge = None

    for item in skill_data.get("debugging_challenges", []):
        if item["challenge_id"] == challenge_id:
            challenge = item
            break

    if challenge is None:
        raise ValueError("Debugging challenge not found.")

    if challenge["score"] is None:
        raise ValueError(
            "Debugging challenge pehle evaluate karo."
        )

    debugging_score = challenge["score"]

    # -----------------------------------------
    # Practice score calculate karo
    # -----------------------------------------

    completed_challenges = [
        item["score"]
        for item in skill_data.get("debugging_challenges", [])
        if item.get("completed") is True
        and item.get("score") is not None
    ]

    if completed_challenges:
        practice_score = round(
            sum(completed_challenges) /
            len(completed_challenges)
        )
    else:
        practice_score = 0

    skill_data["practice_score"] = practice_score
    skill_data["error_challenges_completed"] = len(
        completed_challenges
    )

    # -----------------------------------------
    # Assessment + Practice = Skill Progress
    # -----------------------------------------

    assessment_score = skill_data.get(
        "assessment_score",
        0
    )

    skill_data["progress_percentage"] = round(
        (assessment_score * 0.6) +
        (practice_score * 0.4)
    )

    # -----------------------------------------
    # Update status
    # -----------------------------------------

    progress = skill_data["progress_percentage"]

    if progress == 0:
        skill_data["status"] = "Not Started"

    elif progress < 40:
        skill_data["status"] = "Needs Improvement"

    elif progress < 70:
        skill_data["status"] = "Developing"

    elif progress < 90:
        skill_data["status"] = "Good"

    else:
        skill_data["status"] = "Ready"

    # -----------------------------------------
    # Overall progress refresh
    # -----------------------------------------

    refresh_user_progress(user_state)

    return skill_data


# -----------------------------------------
# Update skill after debugging
# -----------------------------------------

updated_skill = update_skill_after_debugging(
    user_state=user_state,
    skill=debugging_challenge["skill"],
    challenge_id=debugging_challenge["challenge_id"]
)


print("Skill:", debugging_challenge["skill"])
print("Assessment Score:",
      updated_skill["assessment_score"], "%")

print("Practice Score:",
      updated_skill["practice_score"], "%")

print("Skill Progress:",
      updated_skill["progress_percentage"], "%")

print("Status:",
      updated_skill["status"])

print("\nOverall Progress:",
      user_state["overall_progress"], "%")

print("Readiness:",
      user_state["readiness_percentage"], "%")
```

```python
from datetime import datetime
import json
import torch


def save_current_progress_snapshot(user_state):

    timestamp = datetime.now().isoformat()

    if "progress_history" not in user_state:
        user_state["progress_history"] = []

    snapshot = {
        "timestamp": timestamp,
        "overall_progress": user_state["overall_progress"],
        "readiness_percentage": user_state["readiness_percentage"],
        "skills": {}
    }

    for skill, data in user_state["skill_progress"].items():
        snapshot["skills"][skill] = {
            "progress_percentage": data["progress_percentage"],
            "assessment_score": data["assessment_score"],
            "practice_score": data["practice_score"],
            "assessments_completed": data["assessments_completed"],
            "error_challenges_completed": data[
                "error_challenges_completed"
            ],
            "status": data["status"]
        }

    user_state["progress_history"].append(snapshot)

    return snapshot


def refresh_mistral_analysis(user_state):

    current_state = {}

    for skill, data in user_state["skill_progress"].items():
        current_state[skill] = {
            "progress": data["progress_percentage"],
            "assessment_score": data["assessment_score"],
            "practice_score": data["practice_score"],
            "assessments_completed": data["assessments_completed"],
            "error_challenges_completed": data[
                "error_challenges_completed"
            ],
            "status": data["status"]
        }

    prompt = f"""
You are an AI career preparation advisor.

Target Role:
{user_state["target_role"]}

Preparation Timeline:
{user_state.get("timeline_months")} months

LATEST USER STATE:
{json.dumps(current_state, indent=2)}

Previous Analysis:
{user_state.get("current_skill_analysis", "None")}

Re-analyze the user using ONLY the latest state.

Identify:
- Strong skills
- Weak skills
- Skills that improved
- Skills that still need practice
- Next priority
- Preparation focus

Do not invent any score.

Return exactly:

STRONG_SKILLS:
- ...

WEAK_SKILLS:
- ...

IMPROVED_SKILLS:
- ...

NEXT_PRIORITY:
...

PREPARATION_FOCUS:
...

OVERALL_ANALYSIS:
...
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=7000
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=1000,
            temperature=0.4,
            top_p=0.9,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    analysis = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    ).strip()

    user_state["current_skill_analysis"] = analysis

    return analysis


# -----------------------------------------
# Save latest progress
# -----------------------------------------

snapshot = save_current_progress_snapshot(
    user_state
)


# -----------------------------------------
# Refresh Mistral analysis
# -----------------------------------------

latest_analysis = refresh_mistral_analysis(
    user_state
)


print("Progress Snapshot Saved.")
print("Overall Progress:",
      user_state["overall_progress"], "%")

print("Readiness:",
      user_state["readiness_percentage"], "%")

print("\nUPDATED MISTRAL ANALYSIS:\n")
print(latest_analysis)
```

```python
def generate_next_action(user_state):

    skill_data = {}

    for skill, data in user_state["skill_progress"].items():
        skill_data[skill] = {
            "progress": data["progress_percentage"],
            "assessment_score": data["assessment_score"],
            "practice_score": data["practice_score"],
            "assessments_completed": data["assessments_completed"],
            "error_challenges_completed": data[
                "error_challenges_completed"
            ],
            "status": data["status"]
        }

    prompt = f"""
You are an AI career preparation decision engine.

Target Role:
{user_state["target_role"]}

Current Overall Progress:
{user_state["overall_progress"]}%

Current Readiness:
{user_state["readiness_percentage"]}%

Current Skill State:
{json.dumps(skill_data, indent=2)}

Latest Mistral Analysis:
{user_state.get("current_skill_analysis", "Not available")}

Choose the ONE most useful next action for the user.

Possible actions:

ASSESSMENT
Use when the user needs more knowledge evaluation.

DEBUGGING
Use when practical problem-solving or debugging needs improvement.

LEARNING_RESOURCE
Use when the user needs to learn/revise the concept before another assessment.

REVISION
Use when the skill is already strong but needs reinforcement.

READY
Use only when the user's preparation is sufficiently strong.

Do not invent data.

Return EXACTLY:

ACTION:
<ASSESSMENT / DEBUGGING / LEARNING_RESOURCE / REVISION / READY>

SKILL:
<exact skill name>

REASON:
<short explanation>

INSTRUCTION:
<what the user should do next>
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=7000
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=700,
            temperature=0.3,
            top_p=0.9,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    result = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    ).strip()

    user_state["next_action"] = result

    return result


# Generate next action
next_action = generate_next_action(user_state)

print(next_action)
```
